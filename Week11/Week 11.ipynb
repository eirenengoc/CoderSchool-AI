```python
"""
Introduction:
Các bạn vào trang web sau: https://quickdraw.withgoogle.com/
Các bạn hãy chơi thử game trong vòng 1-2 phút
Đây là game QuickDraw, được phát triển dựa trên bộ dataset nổi tiếng cùng tên của Google. Các bạn có thể vào link sau
https://quickdraw.withgoogle.com/data
để có cái nhìn tổng quát về bộ dataset này. Các bạn có thể click vào 1 icon bất kì để xem các bản vẽ của category tương ứng

Pre-requisites:
Các bạn hãy vào link sau đây
https://console.cloud.google.com/storage/browser/quickdraw_dataset/full/numpy_bitmap?pageState=(%22StorageObjectListTable%22:(%22f%22:%22%255B%255D%22))&prefix=&forceOnObjectsSortingFiltering=false
và tự chọn cho mình 10 categories mà các bạn thích (hint: Đừng chọn category nào quá phức tạp)
Sau đó các bạn hãy download các file tương ứng của các categories này (click vào mũi tên đi xuống ở dòng tương ứng)
Các file được download sẽ có tên như sau: full_numpy_bitmap_{category}.npy. Sau đó các bạn hãy để các file .npy này
vào chung với folder của script python này. Giờ các bạn có thể chuyển sang phần tiếp theo - phần lập trình
"""

# Import libraries:
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import random

# List of categories:
category_files = {
    "mona_lisa": "/Users/ngocta/Desktop/CoderSchool-AI/Week11/Doodle Library/full_numpy_bitmap_The Mona Lisa.npy",
    "apple": "/Users/ngocta/Desktop/CoderSchool-AI/Week11/Doodle Library/full_numpy_bitmap_apple.npy",
    "birthday_cake": "/Users/ngocta/Desktop/CoderSchool-AI/Week11/Doodle Library/full_numpy_bitmap_birthday cake.npy",
    "broom": "/Users/ngocta/Desktop/CoderSchool-AI/Week11/Doodle Library/full_numpy_bitmap_broom.npy",
    "cellphone": "/Users/ngocta/Desktop/CoderSchool-AI/Week11/Doodle Library/full_numpy_bitmap_cell phone.npy",
    "dog": "/Users/ngocta/Desktop/CoderSchool-AI/Week11/Doodle Library/full_numpy_bitmap_dog.npy",
    "grape": "/Users/ngocta/Desktop/CoderSchool-AI/Week11/Doodle Library/full_numpy_bitmap_grapes.npy",
    "pig": "/Users/ngocta/Desktop/CoderSchool-AI/Week11/Doodle Library/full_numpy_bitmap_pig.npy",
    "pizza": "/Users/ngocta/Desktop/CoderSchool-AI/Week11/Doodle Library/full_numpy_bitmap_pizza.npy"
}

# Define function to process each category:
def process_category(file_path, category_name, visualize = True):
    
    # STEP 1: 
    # Load image:
    images = np.load(file_path).astype(np.float32)
    shape = images.shape
    # Split to train and test sets:
    train_images = images[:-10]
    test_images = images[-10:]

    # STEP 2: 
    # Calculate the average image from the train dataset:
    average_image = np.mean(train_images, axis = 0) 
    # Reshape from vector 784 to image size 28x28 to display:
    average_image_reshaped = average_image.reshape(28, 28)
    # Display the average image:
    if visualize:
        plt.imshow(average_image_reshaped, cmap='gray')
        plt.title(f"Average Image: {category_name}")
        plt.axis('off')
        plt.show()
        
    # STEP 4: 
    # Get a test image from a random index from 1-9:
    index = random.randint(1, 9)
    test_image = test_images[index]
    # Calculate Dot product between a test image and the average image:
    dot_product = np.dot(test_image, average_image)
    print(f"Dot product (test image index {index}): {dot_product:,.2f}")
    # Calculate average dot product across all test images
    all_dot_products = [np.dot(img, average_image) for img in test_images]
    average_dot_product = np.mean(all_dot_products)
    # Descriptive conclusion
    ratio = dot_product / average_dot_product
    if ratio < 0.8:
        conclusion = "Low similarity: Test image very different from average image."
    elif ratio > 1.2:
        conclusion = "High similarity: Test image very similar to average image."
    else:
        conclusion = "Moderate similarity: Test image fairly similar to average image."

    return {
        "Category": category_name,
        "Shape": shape,
        "Test Index": index,
        "Dot Product": dot_product,
        "Avg Dot Product": average_dot_product,
        "Conclusion": conclusion
    }

# Loop through each category:
results = []
for category, path in category_files.items():
    result = process_category(path, category)
    results.append(result)

# Convert to DataFrame for display
df_results = pd.DataFrame(results)
print(df_results)
```
