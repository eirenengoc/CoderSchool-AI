```python
# WEEK 4A
# Import data:
import pandas as pd
data = pd.read_csv(r'/Users/ngocta/Desktop/CoderSchool-AI/Week4/Week4_1_data.csv', sep = ',')
print(data)
print(data.dtypes)

# Task 1: Thống kê số lượng khách hàng theo học vấn (High school, master,...)
education_summary = (
    data.groupby('educational_level')
    .size()
    .reset_index(name = "total_size")
    .sort_values(by = "total_size", ascending = False)
)
print(education_summary)
```

```python
# Task 2: In ra 20 khách hàng có thu nhập cá nhân (annual_income) cao nhất
top_20 = (
    data.sort_values(by = 'annual_income', ascending = False)
    .head(20)
    .reset_index(drop = True)
)
print(top_20)
```

```python
# Task 3: In ra những khách hàng sinh sau năm 1960 và thu nhập cá nhân trên $50,000/năm
filtered_list3 = (
    data[(data['year_of_birth'] >1960) & (data['annual_income'] >50000)]
    .reset_index(drop = True)
)
print(filtered_list3)
```

```python
# Task 4: Kết hợp điều kiện của Task 2 & 3
filtered_list4 = (
    data[(data['year_of_birth'] > 1960) & (data['annual_income']>50000)]
    .sort_values(by = 'annual_income', ascending = False)
    .reset_index(drop = True)
    .head(20)
)
print(filtered_list4)
```

```python
# Task 5: In ra những khách hàng có tình trạng hôn nhân là đã kết hôn or đã ly hôn
data['marital_status'].unique()
filtered_list5 = data[data['marital_status'].isin(['Married','Divorced'])].reset_index(drop= True)
print(filtered_list5)
```

```python
# Task 6: Tính mức thu nhập trung bình theo trình độ học vấn
filtered_list6 = (
    data.groupby('educational_level')
    ['annual_income'].mean().round(1)
    .reset_index(name = 'avg_annual_income')
)
print(filtered_list6)
```

```python
# Task 7: Tính mức thu nhập trung bình theo trình độ học vấn + tình trạng hôn nhân
filtered_list7 = (
    data.groupby(['educational_level', 'marital_status'])['annual_income'].mean().round(1)
    .reset_index(name = 'avg_annual_income')
    .sort_values(['educational_level', 'marital_status','avg_annual_income'], ascending = False)
    .reset_index(drop = True)
)
print(filtered_list7)
```

```python
# Bài 3b: Nhập input đầu vào là 2 thông số, chiều dài & chiều rộng của 1 bông hoa bất kỳ 
# (theo đơn vị mm). Xác định xem bông hoa đấy thuộc loại hoa nào (Iris setosa, Iris virginica hay Iris versicolor). 
# Loại của bông hoa được xác định là class của
# a) Điểm có tọa độ gần nó nhất. Nếu có 2 điểm thuộc 2 loại hoa khác nhau cùng 
# nằm gần nhất với input, thì xác định sang điểm kế tiếp
length = np.random.choice(data['SepalLengthCm'])
width = np.random.choice(data['SepalWidthCm'])
data['Distance'] = np.sqrt((data['SepalLengthCm']-length)**2 + (data['SepalWidthCm']- width)**2)
sorted_data = data.sort_values(by = 'Distance', ascending = True).reset_index(drop = True)

#Calculate the closest distance:
closest_distance = sorted_data['Distance'].iloc[0]

#Get all the flower types that having the same closest distance:
sorted_results = sorted_data[sorted_data['Distance'] == closest_distance]
print(sorted_results)

# Check if there's only 1 flower type in the list:
if sorted_results['Species'].nunique() == 1:
    print(f"This flower is a {sorted_results['Species'].iloc[0]}")
# If there're more than 1 flower type:
else:
    type_count = sorted_results['Species'].value_counts()
    # Check if there're a tie in count between different flower type
    if len(type_count.unique()) == 1:
        #Return the default one (1st row):
        print(f"This flower is a {sorted_results['Species'].iloc[0]}")
    else:
        #Return the flower type with the max count
        print(f"This flower is a {type_count.idxmax()}")
```

```python
# b) 7 điểm gần nó nhất (Vote đa số). Nếu không xác định được class chiếm đa số,
# in ra bông hoa này ko thuộc class nào

# Get the 7 closest points:
sorted_data_b = sorted_data.head(7)
# Count them by flower type
type_count_b = sorted_data_b['Species'].value_counts()
print(type_count_b)

# Check if there's a tie between different flower type:
if (type_count_b == type_count_b.max()).sum() > 1:
    print(f"This flower doesn't belong to any class")
else:
    # If not, return the flower type with the max count
    print(f"This flower belongs to {type_count_b.idxmax()}")
```

```python
# c) Lấy tọa độ của input là tâm của 1 đường tròn, có bán kính 2 mm. Vote đa số
# dựa trên tất cả các điểm nằm trên và trong đường tròn. Nếu có 2 class có số
# điểm bằng nhau và đồng cao nhất, in ra thông báo bông hoa này ko thuộc class nào.

#Extract data list with all the points in & on the radius:
sorted_data_c = sorted_data[sorted_data['Distance'] <= 0.2]
type_count_c = sorted_data_c['Species'].value_counts()
print(type_count_c)

#Check if there's only 1 flower type
if type_count_c.nunique() == 1:
    print(f"This flower is a {type_count_c.index[0]}")
else:
    #If there's a tie count between different flower type:
    if (type_count_c == type_count_c.max()).sum() > 1:
        print(f"This flower doesnt belong to any class")
    # If there's a flower type with the highest count:
    else:
        print(f"This flower is a {type_count_c.idxmax()}")
```
