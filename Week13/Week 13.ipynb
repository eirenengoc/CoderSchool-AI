```python
import torch
torch.manual_seed(100)

def activation_func(x):
    # Using ReLU activation
    return torch.relu(x)

def softmax(x):
    # Subtracting max to avoid data overflow:
    e_x = torch.exp(x - torch.max(x))
    return e_x / torch.sum(e_x, dim=1, keepdim=True)

# Define the size of each layer in the network
num_input = 784  # Number of nodes in input layer (28x28)
num_hidden_1 = 128  # Number of nodes in hidden layer 1
num_hidden_2 = 256  # Number of nodes in hidden layer 2
num_hidden_3 = 128  # Number of nodes in hidden layer 3
num_classes = 10  # Number of nodes in output layer

# Randomize input
input_data = torch.randn((1, num_input))
# Randomize weights for inputs to hidden layer 1
W1 = torch.randn(num_input, num_hidden_1)
# Randomize weights for hidden layer 1 to hidden layer 2
W2 = torch.randn(num_hidden_1, num_hidden_2)
# Randomize weights for hidden layer 2 to hidden layer 3
W3 = torch.randn(num_hidden_2, num_hidden_3)
# Randomize weights for hidden layer 3 to output layer
W4 = torch.randn(num_hidden_3, num_classes)

# Calculate Bias for hidden and output layers
B1 = torch.randn((1, num_hidden_1))
B2 = torch.randn((1, num_hidden_2))
B3 = torch.randn((1, num_hidden_3))
B4 = torch.randn((1, num_classes))

# Forward pass
out1 = activation_func(torch.matmul(input_data, W1) + B1)
out2 = activation_func(torch.matmul(out1, W2) + B2)
out3 = activation_func(torch.matmul(out2, W3) + B3)
logits = torch.matmul(out3, W4) + B4
result = softmax(logits)

print("Output:", result)
print("Sum of output probabilities:", torch.sum(result).item())
```
